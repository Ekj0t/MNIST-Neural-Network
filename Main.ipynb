### MNIST Neural Network

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
data = pd.read_csv('Dataset/mnist_train.csv')
```

```python
# We don't want to use pandas dataframes for our neural network, so we will convert the data to numpy arrays
data = np.array(data)
# m is the number of rows in the data, n is the number of columns in the data
m, n = data.shape
# Suffle the data to prevent overfitting
np.random.shuffle(data)

data_dev = data[0:1000].T
Y_dev = data_dev[0]
X_dev = data_dev[1:n]

data_train = data[1000:m].T
Y_train = data_train[0]
X_train = data_train[1:n]
```
